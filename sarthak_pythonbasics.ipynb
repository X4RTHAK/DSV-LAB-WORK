```python
# Q7.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

print(my_abs(5))
print(my_abs(-7))
print(my_abs(0))
```

```python
# Q7.2
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

try:
    print(my_abs('Hi'))
except TypeError as e:
    print('TypeError:', e)
# A string cannot be compared to 0 using <.
```

```python
# Q7.3
def print_abs(val):
    if val < 0:
        print(-val)
    else:
        print(val)

x = print_abs(-2.7)
print(x)  # None, because function has no return statement.
```

```python
# Q7.4
def inc_val(val):
    val = val + 1

x = 7
inc_val(x)
print(x)  # 7: rebinding a local parameter doesn't reassign x.
```

```python
# Q7.5
def swap_message(val1, val2):
    val1, val2 = val2, val1
    print('Inside function:', val1, val2)

x, y = 6, 3
swap_message(x, y)
print('Original variables:', x, y)
```

```python
# Q8.1
def my_abs(val):
    if val < 0:
        return 0 - val
    return val

try:
    print(val)
except NameError as e:
    print('NameError:', e)
# val only exists as a local parameter inside my_abs.
```

```python
# Q8.2
counter = 0
def show_counter():
    print(counter)

show_counter()
```

```python
# Q8.3
counter = 0
def increment_counter():
    counter = counter + 1

try:
    increment_counter()
except UnboundLocalError as e:
    print('UnboundLocalError:', e)
# Assignment makes counter local, so its value isn't defined yet.
```

```python
# Q8.4
counter = 0
def increment_counter():
    global counter
    counter = counter + 1

increment_counter()
print('Counter:', counter)
```
