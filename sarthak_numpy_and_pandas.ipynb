```python
# Q35
import numpy as np
rng = np.random.default_rng(42)
marks = rng.integers(20, 101, (12, 4))
student_avg = np.mean(marks, axis=1)
test_mean = np.mean(marks, axis=0)
test_std = np.std(marks, axis=0)
all_pass = np.all(marks >= 40, axis=1)
print('Marks (rows = students 1 to 12):', marks)
print('Student averages:', student_avg)
print('Test means:', test_mean)
print('Test std:', test_std)
print('Students passing all tests:', np.sum(all_pass))
print('Students average >=75:', np.where(student_avg >= 75)[0] + 1)
print('Test with greatest variation:', np.argmax(test_std) + 1)
```

```python
# Q36
import numpy as np
a = np.arange(21, 46).reshape(5, 5)
center = a[1:4, 1:4]
t = center.T
product = np.dot(center, t)
max_value = np.max(product)
print('Original (5x5):', a)
print('Center (3x3):', center)
print('Transpose (3x3):', t)
print('Product (3x3):', product)
print('Row sums:', np.sum(product, axis=1))
print('Maximum:', max_value)
print('Positions (0-based):', np.argwhere(product == max_value))
```

```python
# Q37
import numpy as np
quantities = np.array([[12, 18, 9, 15], [20, 17, 11, 19], [22, 15, 14, 17], [16, 20, 12, 18], [19, 21, 16, 20], [18, 16, 10, 14]])
prices = np.array([150, 240, 90, 320])
daily_revenue = np.sum(quantities * prices, axis=1)
product_quantity = np.sum(quantities, axis=0)
product_revenue = np.sum(quantities * prices, axis=0)
new_prices = np.where(product_quantity > 100, prices * 0.9, prices)
print('Daily revenue:', daily_revenue)
print('Revenue per product:', product_revenue)
print('Highest revenue day:', np.argmax(daily_revenue) + 1)
print('Best-selling product by quantity:', np.argmax(product_quantity) + 1)
print('Total quantities:', product_quantity)
print('Original prices:', prices)
print('Discounted prices:', new_prices)
print('Original total revenue:', daily_revenue.sum())
print('New total revenue:', np.sum(quantities * new_prices))
```

```python
# Q38
import numpy as np
rng = np.random.default_rng(42)
a = rng.normal(50, 12, 100)
outside = np.sum((a < 30) | (a > 70))
clean = np.where(a < 30, 30, np.where(a > 70, 70, a))
print('Outside 30–70:', outside)
print('Original statistics:', np.mean(a), np.var(a), np.std(a), np.min(a), np.max(a))
print('Cleaned statistics:', np.mean(clean), np.var(clean), np.std(clean), np.min(clean), np.max(clean))
print('Clipping limits extreme values; variance and std generally decrease, and mean can shift')
```

```python
# Q39
import pandas as pd
branch_a = pd.Series({'Q1': 120, 'Q2': 145, 'Q4': 180})
branch_b = pd.Series({'Q2': 110, 'Q3': 135, 'Q4': 155})
branch_c = pd.Series({'Q1': 95, 'Q3': 125, 'Q4': 140})
ordinary = branch_a + branch_b + branch_c
common = branch_a.index.intersection(branch_b.index).intersection(branch_c.index)
zero_filled = branch_a.add(branch_b, fill_value=0).add(branch_c, fill_value=0)
print('Ordinary addition:', ordinary)
print('Common quarters:', common.tolist())
print('Combined earnings in common quarters:', ordinary.loc[common])
print('Absent records treated as zero:', zero_filled)
print('Ordinary sum needs all branches; zero-filled sum counts recorded earnings only')
```

```python
# Q40
import pandas as pd
scores = {'R101': 78, 'R102': 65, 'R104': 92, 'R105': 48}
required = ['R105', 'R103', 'R101', 'R106', 'R104', 'R102']
s = pd.Series(scores, index=required)
missing = s[s.isnull()]
filtered = s[s >= 70]
result_dict = filtered.to_dict()
restored = pd.Series(result_dict)
print('Series in required order:', s)
print('Missing records:', missing.index.tolist())
print('Valid scores >= 70:', filtered)
print('Dictionary:', result_dict)
print('Reconstructed:', restored)
print('Same labelled values:', filtered.equals(restored))
print('Missing scores mean unknown, not zero')
```
