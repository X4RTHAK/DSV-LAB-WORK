```python
# Q28
import pandas as pd
s = pd.Series([90, 75, 82], index=['Ravi', 'Aman', 'Neha'])
d = s.to_dict()
s2 = pd.Series(d)
print('Dictionary:', d)
print('New Series:')
print(s2)
print('Same labelled data:', s.equals(s2))
```

```python
# Q29
import pandas as pd
sales = pd.Series([450, 720, 390, 850], index=['Pen', 'Bag', 'Book', 'Bottle'])
print('Values:', sales.values)
print('Index:', sales.index)
print('Bag sales:', sales['Bag'])
print('Sales above 500:')
print(sales[sales > 500])
print('Book exists:', 'Book' in sales)
print('Dictionary:', sales.to_dict())
```

```python
# Q30
import pandas as pd
data = {'A': 15, 'B': 35, 'C': 25}
s = pd.Series(data, index=['C', 'A', 'D', 'B'])
s.name = 'Observations'
s.index.name = 'Category'
print('(a) Index order:', list(s.index))
print('(b) NaN label: D')
print('(c) Values above 20:')
print(s[s > 20])
print('(d) Non-missing values:')
print(pd.notnull(s))
print('(e) Series name:', s.name)
print('Index name:', s.index.name)
```
